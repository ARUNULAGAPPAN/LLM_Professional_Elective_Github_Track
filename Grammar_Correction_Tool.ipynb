```python
# 1. Load the "Large" (Smarter) AI Model
from transformers import pipeline
import gradio as gr

print("Downloading the 'Large' Smart Model... (This takes about 2 minutes)")
# We use 'pipeline' which handles all the complexity for us automatically
corrector = pipeline("text2text-generation", "pszemraj/flan-t5-large-grammar-synthesis")
print("✅ Smart Model Loaded!")

# 2. Define the Function
def smart_fix(text):
    if not text:
        return ""

    # This specific model works best without any "grammar:" prefix
    # We ask it to generate up to 512 tokens (words)
    results = corrector(text, max_length=512)

    # Extract the text from the result
    corrected_text = results[0]['generated_text']
    return corrected_text

# 3. Create the Interface
interface = gr.Interface(
    fn=smart_fix,
    inputs=gr.Textbox(lines=3, placeholder="Type: 'what was your time now ?'"),
    outputs="text",
    title="🧠 Smart AI Grammar Corrector (Large Model)",
    description="This uses a larger, smarter AI (T5-Large) that understands context better than the base model.",
    examples=[
        ["what was your time now ?"],
        ["i can has cheezburger"],
        ["She don't like playing soccer."]
    ]
)

# 4. Launch
interface.launch(share=True)
```
